```python
def categorize_athlete(row):
  if row['height_cm'] < 175 and row['weight_kg'] < 70:
    return 'Lightweight'
  elif row['height_cm'] < 185 or row['weight_kg'] <= 80:
    return 'Middleweight'
  else:
    return 'Heavyweight'
bios_new['category'] = bios.apply(categorize_athlete, axis = 1)
bios_new.head()
```
